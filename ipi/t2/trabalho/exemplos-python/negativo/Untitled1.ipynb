```python
import numpy as np               # para trabalhar com matrizes de forma eficiente
import matplotlib.pyplot as plt  # para plotagem
```

```python
img = plt.imread("breast_digital_Xray.tif") # retorna um array numpy
print(img.shape) 
```

## Negativo de imagem acromática

 Fórmula: g(x,y) = 255 - f(x,y)  # funciona por broadcast numpy
 (571, 482)
 (1,     1)

```python
neg = 255 - img  #numpy
print(neg.shape)
```

```python
fig, axs = plt.subplots(1, 2)
axs[0].set_title("Original")
axs[0].axis("off")
axs[0].imshow(img, cmap="gray")

axs[1].set_title("Negativo")
axs[1].axis("off")
axs[1].imshow(neg, cmap="gray")
```

## Ajuste de Brilho em Imagens Acromáticas

```python
c = 100
imgMaisBrilho = img.astype(np.float64) + c

# trunca valores saturados
imgMaisBrilho[imgMaisBrilho > 255] = 255
imgMaisBrilho = imgMaisBrilho.astype(np.uint)
print(imgMaisBrilho)
```

```python
fig, axs = plt.subplots(1, 2)
axs[0].set_title("Original")
axs[0].axis("off")
axs[0].imshow(img, cmap="gray")

axs[1].set_title("+Brilho")
axs[1].axis("off")
axs[1].imshow(imgMaisBrilho, cmap="gray")
```

```python
c = 30
imgMenosBrilho = img.astype(np.float64) - c

# trunca valores saturados
imgMenosBrilho[imgMenosBrilho < 0] = 0
imgMenosBrilho = imgMenosBrilho.astype(np.uint)
print(imgMenosBrilho)
```

```python
fig, axs = plt.subplots(1, 2)
axs[0].set_title("Original")
axs[0].axis("off")
axs[0].imshow(img, cmap="gray")

axs[1].set_title("-Brilho")
axs[1].axis("off")
axs[1].imshow(imgMenosBrilho, cmap="gray")
```
